```python
# CREATES SPARK SESSION

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Sales Analysis") \
    .getOrCreate()

print("Spark version:", spark.version)
```

```python
# CHECKING FILE

import os

path = "/content/drive/MyDrive/Sales_Data.csv"

print(path)
```

```python
# READ FILE

sales_df = spark.read.csv(
    path,
    header = True,
    inferSchema = True
)

sales_df.show(7)
```

```python
# CALCULATE PRODUCTS SOLD IN EACH CATEGORY

from pyspark.sql.functions import sum

grouped_data = sales_df.groupBy("PRODUCTLINE")

summed_data = grouped_data.agg(sum("QUANTITYORDERED").alias("TOTAL_PRODUCT_SOLD"))


result = summed_data.orderBy("PRODUCTLINE")

result.show()
```
